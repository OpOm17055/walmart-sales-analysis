```python
import pandas as pd
```

```python

df = pd.read_csv('Walmart.csv', encoding_errors='ignore')
```

```python
df.head()
```

```python
df.info()
print(df.isnull().sum())
print('Duplicates:', df.duplicated().sum())
df.head()
```

```python
df = df.drop_duplicates()
```

```python
df = df.dropna()
```

```python
df.columns = df.columns.str.lower()
```

```python
df['unit_price'] = df['unit_price'].str.replace('$', '', regex=False).astype(float)
```

```python
df['quantity'] = df['quantity'].astype(int)
```

```python
df['total'] = df['unit_price'] * df['quantity']
```

```python
print(df.shape)
df.info()
```

```python
df.to_csv('walmart_clean.csv', index=False)
files.download('walmart_clean.csv')
```

```python
df['profit'] = df['total'] * df['profit_margin']
```

```python
df['date'] = pd.to_datetime(df['date'], format='%d/%m/%y')
```

```python
df['year'] = df['date'].dt.year
```

```python
df['month_name'] = df['date'].dt.month_name()
```

```python
df['day_name'] = df['date'].dt.day_name()
```

```python
df['hour'] = df['time'].str.split(':').str[0].astype(int)
```

```python
df['shift'] = pd.cut(df['hour'], bins=[-1, 11, 17, 24],labels=['Morning', 'Afternoon', 'Evening']).astype(str)
```

```python
df[['date', 'year', 'month_name', 'day_name', 'hour', 'shift', 'total', 'profit']].head()
```

```python
print(df.shape)
print(df['year'].value_counts().sort_index())
print(df['shift'].value_counts())
print('Revenue:', round(df['total'].sum()), '| Profit:', round(df['profit'].sum()))
```

```python
df.to_csv('walmart_enriched.csv', index=False)
files.download('walmart_enriched.csv')
```
